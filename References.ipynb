# Model and Preprocessing Technical Manual

This manual details the data preparation pipelines and machine learning classifiers applied during the development of the Employee Performance Analysis project.

## 1. Feature Preprocessing Pipeline

* **Data Cleaning:** Missing values in numerical attributes are handled programmatically using median imputation to prevent data distortion from potential outliers.
* **Categorical Encoding (`LabelEncoder`):** String/Object categories (such as `EmpDepartment`, `EmpJobRole`, `Gender`, and `OverTime`) are transformed into numerical integer encodings via Scikit-Learn's `LabelEncoder` to ensure compatibility with numerical modeling frameworks.
* **Feature Scaling (`StandardScaler`):** Features are standardized by centering and scaling to unit variance using:
  $$z = \frac{x - \mu}{\sigma}$$
  This step balances the weight distribution among variables with disparate ranges (e.g., `Age` vs. `EmpHourlyRate`).

## 2. Supervised Learning Algorithms Evaluated

To find the optimal predictive structure for tracking performance ratings, the following classifiers were built and assessed using a train-test split structure:

1. **Logistic Regression (`LogisticRegression`):** Implemented with a `class_weight='balanced'` parameter to automatically handle implicit class frequencies.
2. **Decision Tree Classifier (`DecisionTreeClassifier`):** Utilized as a baseline non-linear structure using balanced weights to establish feature split importance thresholds.
3. **Random Forest Classifier (`RandomForestClassifier`):** Built as an ensemble of 100 parallel estimator trees to improve generalizability and reduce overfitting.
4. **AdaBoost Classifier (`AdaBoostClassifier`):** Utilized sequential boosting techniques to iteratively reduce residual estimation errors.
5. **Gradient Boosting Classifier (`GradientBoostingClassifier`):** Implemented forward-stage additive models to capture deep non-linear interaction patterns.
6. **K-Neighbors Classifier (`KNeighborsClassifier`):** A distance-based instance classifier used to evaluate similarity maps across employee attribute sets.

## 3. Evaluation Protocol
Models are rigorously judged via multi-class evaluation scores containing:
* Classification Accuracy
* Precision Score
* Recall / Sensitivity
* $F_1\text{-Score}$
* Confusion Matrix analysis

# Explanatory Domain Materials & Insights

This file contains contextual business logic, domain descriptions, and insights concerning how human resource metrics align with employee performance outcomes.

## 1. Target Target Distribution Realities

The target attribute `PerformanceRating` follows an imbalanced distribution across the corporate dataset:
* **Rating 3 (Excellent):** 874 Employees
* **Rating 2 (Good):** 194 Employees
* **Rating 4 (Outstanding):** 132 Employees

Because Level 3 accounts for the vast majority of records, specialized balancing techniques (such as setting model class-weights to `balanced`) are necessary to ensure that Level 2 and Level 4 scores are predicted with high recall and precision.

## 2. Strategic Insights from Exploratory Analysis

* **Key Correlation Drivers:** Exploratory analysis shows that features like `EmpEnvironmentSatisfaction`, `EmpLastSalaryHikePercent`, and `EmpWorkLifeBalance` exhibit significant correlations with high `PerformanceRating` scores.
* **Department-Specific Performance:** Different operational sectors (such as Development, Sales, and Research & Development) exhibit varying baseline average ratings, implying that evaluation standards often shift based on operational frameworks.
* **The Overtime and Balance Trade-off:** Tracking the interplay between `OverTime`, `EmpJobSatisfaction`, and the overall target metrics uncovers core indicators of burn-out vs. hyper-productivity within specific departments.
