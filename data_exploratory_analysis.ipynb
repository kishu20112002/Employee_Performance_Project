# Exploratory Data Analysis (EDA)

## Objective

This notebook focuses on:

- Employee data visualization
- Distribution analysis
- Department analysis
- Employee behavior patterns
- Correlation analysis

EDA helps identify important factors affecting employee performance.

```python
# Import Required Libraries

import pandas as pd
import numpy as np

import matplotlib.pyplot as plt
import seaborn as sns

import warnings
warnings.filterwarnings('ignore')
```

```python
# Load Dataset

df = pd.read_csv(
    "/Users/apple/code/Datamites Project/Employee_Performance_Project/Data/Raw/INX_Future_Inc_Employee_Performance_CDS_Project2_Data_V1.8.csv"
)

df.head()
```

```python
# Employee Performance Rating Distribution

plt.figure(figsize=(8,5))

sns.countplot(
    x='PerformanceRating',
    data=df
)

plt.title(
    "Employee Performance Rating Distribution"
)

plt.show()
```

```python
# Department Wise Employee Distribution

plt.figure(figsize=(10,5))

sns.countplot(
    y='EmpDepartment',
    data=df
)

plt.title(
    "Department Wise Employee Distribution"
)

plt.show()
```

```python
# Gender Distribution

plt.figure(figsize=(6,5))

sns.countplot(
    x='Gender',
    data=df
)

plt.title(
    "Gender Distribution"
)

plt.show()
```

```python
# Education Background Distribution

plt.figure(figsize=(10,5))

sns.countplot(
    y='EducationBackground',
    data=df
)

plt.title(
    "Education Background Distribution"
)

plt.show()
```

```python
# Age Distribution

plt.figure(figsize=(8,5))

sns.histplot(
    df['Age'],
    kde=True
)

plt.title(
    "Age Distribution"
)

plt.show()
```

```python
# Employee Hourly Rate Distribution

plt.figure(figsize=(8,5))

sns.histplot(
    df['EmpHourlyRate'],
    kde=True
)

plt.title(
    "Employee Hourly Rate Distribution"
)

plt.show()
```

```python
# Job Satisfaction Distribution

plt.figure(figsize=(8,5))

sns.countplot(
    x='EmpJobSatisfaction',
    data=df
)

plt.title(
    "Employee Job Satisfaction"
)

plt.show()
```

```python
# Work Life Balance Distribution

plt.figure(figsize=(8,5))

sns.countplot(
    x='EmpWorkLifeBalance',
    data=df
)

plt.title(
    "Employee Work Life Balance"
)

plt.show()
```

```python
# Correlation Heatmap

plt.figure(figsize=(18,10))

sns.heatmap(
    df.corr(numeric_only=True),
    annot=True,
    cmap='coolwarm'
)

plt.title(
    "Correlation Heatmap"
)

plt.show()
```

# EDA Insights

Key observations from Exploratory Data Analysis:

1. Employee performance ratings vary across employees.

2. Department-wise employee distribution differs significantly.

3. Salary-related variables influence employee performance.

4. Work-life balance impacts productivity.

5. Training and experience contribute to employee efficiency.

6. Correlation analysis identified influential variables affecting employee performance.
