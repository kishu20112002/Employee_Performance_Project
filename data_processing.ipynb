# Missing Value Handling

Missing values are checked to ensure dataset quality and consistency.

```python
# Missing Values Check

missing_values = df.isnull().sum()

missing_values[
    missing_values > 0
]
```

# Duplicate Removal

Duplicate records can negatively affect model performance and introduce bias.

Therefore, duplicate observations are removed.

```python
# Duplicate Rows

print(
    "Duplicate Rows:",
    df.duplicated().sum()
)
```

```python
# Remove Duplicate Rows

df = df.drop_duplicates()

print(
    "New Shape:",
    df.shape
)
```

```python
# Data Types

df.dtypes
```

```python
# Numerical and Categorical Columns

numerical_cols = df.select_dtypes(
    include=np.number
).columns

categorical_cols = df.select_dtypes(
    include='object'
).columns

print(
    "Numerical Columns:"
)

print(
    numerical_cols
)

print(
    "\nCategorical Columns:"
)

print(
    categorical_cols
)
```

# Label Encoding

Machine learning models cannot understand text data.

Therefore, categorical columns are converted into numerical form using Label Encoding.

```python
# Label Encoding

label_encoder = LabelEncoder()

for column in categorical_cols:

    df[column] = label_encoder.fit_transform(
        df[column]
    )

# Display Data
df.head()
```

```python
# Feature and Target Variable

X = df.drop(
    'PerformanceRating',
    axis=1
)

y = df[
    'PerformanceRating'
]

print(
    "Feature Shape:",
    X.shape
)

print(
    "Target Shape:",
    y.shape
)
```

```python
# Train Test Split

X_train, X_test, y_train, y_test = train_test_split(

    X,
    y,

    test_size=0.20,

    random_state=42
)

print(
    "Training Shape:",
    X_train.shape
)

print(
    "Testing Shape:",
    X_test.shape
)
```
